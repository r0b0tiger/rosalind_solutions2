```python
#DNA
def base_count(x):
    dict_nuc_count = {}
    for i in x:
        if i in dict_nuc_count:
            dict_nuc_count[i] += 1
        else:
            dict_nuc_count[i] = 1
    ret_ans = str(dict_nuc_count['A']) + ' ' + str(dict_nuc_count['C']) + ' ' + str(dict_nuc_count['G']) + ' ' + str(dict_nuc_count['T'])
    return ret_ans
```
